```python
import os
import glob
import cv2
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files, drive
from ultralytics import YOLO

# 1. Garante montagem do Drive para buscar pesos salvos
drive.mount('/content/drive')

BASE_PATH = "/content/drive/MyDrive/Dio/Desafio_Detecção_Facial"

# 2. Localização automática dos pesos treinados (best.pt)
candidatos_pesos = [
    f"{BASE_PATH}/yolo_runs/friends_yolo_treino/weights/best.pt",
    "runs/detect/train/weights/best.pt",
    "best.pt"
]
caminho_modelo = next((p for p in candidatos_pesos if os.path.exists(p)), None)

if not caminho_modelo:
    buscas = glob.glob(f"{BASE_PATH}/**/best.pt", recursive=True) + glob.glob("**/best.pt", recursive=True)
    caminho_modelo = buscas[0] if buscas else "yolo11n.pt"

print(f"-> Modelo carregado: {caminho_modelo}")
model = YOLO(caminho_modelo)

# 3. Seleção da Imagem de Teste (Upload ou Drive)
print("\nComo deseja selecionar a foto de teste?")
print("1 - Fazer upload direto do computador")
print("2 - Usar foto já salva no Google Drive (friends.jpeg)")
opcao = input("Digite 1 ou 2: ").strip()

if opcao == "1":
    print("\nSelecione a imagem de grupo do seu computador:")
    uploaded = files.upload()
    caminho_imagem = list(uploaded.keys())[0]
else:
    candidatos_img = (
        glob.glob(f"{BASE_PATH}/**/*friend*.jpeg", recursive=True) +
        glob.glob(f"{BASE_PATH}/**/*friend*.jpg", recursive=True) +
        glob.glob(f"{BASE_PATH}/**/*friends*.png", recursive=True)
    )
    if candidatos_img:
        caminho_imagem = candidatos_img[0]
        print(f"Foto localizada no Drive: {caminho_imagem}")
    else:
        raise FileNotFoundError("Nenhuma imagem com 'friend' encontrada no Drive.")

# 4. Inferência Calibrada para Múltiplas Faces em Grupo
# - conf=0.12: Captura faces mesmo com pontuação modesta em grupo
# - iou=0.65: Permite caixas próximas sem supressão mútua
# - imgsz=1024: Preserva detalhes de faces pequenas
print(f"\nProcessando imagem: {caminho_imagem}...")
results = model.predict(
    source=caminho_imagem,
    conf=0.12,
    iou=0.65,
    imgsz=1024,
    save=False,
    verbose=False
)

boxes = results[0].boxes
total_faces = len(boxes)

# 5. Painel de Diagnóstico no Console
print("\n" + "=" * 50)
print(f"   DIAGNÓSTICO: {total_faces} FACE(S) DETECTADA(S)")
print("=" * 50)

if total_faces > 0:
    for i, box in enumerate(boxes):
        cls_id = int(box.cls[0])
        nome_ator = model.names[cls_id].replace("_", " ").title()
        score = float(box.conf[0]) * 100
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        print(f"[{i+1}] {nome_ator:<20} | Confiança: {score:.1f}% | Caixa: ({x1}, {y1}) -> ({x2}, {y2})")
else:
    print("Nenhuma face identificada com o limiar atual.")
print("=" * 50)

# 6. Renderização Gráfica no Google Colab
imagem_anotada = results[0].plot(line_width=2, font_size=0.7)
imagem_rgb = cv2.cvtColor(imagem_anotada, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 10))
plt.imshow(imagem_rgb)
plt.axis("off")
plt.title(f"Detecção Multi-Face em Grupo - Total: {total_faces} pessoas", fontsize=15, pad=12)

saida_png = "/content/resultado_grupo_final.png"
plt.savefig(saida_png, bbox_inches='tight', dpi=300)
plt.show()

print(f"\nImagem com as caixas salva em: {saida_png}")
```
